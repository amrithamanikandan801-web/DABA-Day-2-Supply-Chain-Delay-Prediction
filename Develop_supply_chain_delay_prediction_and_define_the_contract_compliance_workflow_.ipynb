```python
import pandas as pd
import numpy as np
import random

np.random.seed(42)
random.seed(42)

print("Libraries imported successfully!")
```

```python
n = 500

vendors = [
    "Alpha Logistics",
    "Beta Suppliers",
    "Gamma Industries",
    "Delta Traders",
    "Epsilon Exports",
    "Zeta Manufacturing",
    "Omega Supplies",
    "Prime Logistics"
]

origins = [
    "Chennai",
    "Coimbatore",
    "Bangalore",
    "Hyderabad",
    "Mumbai",
    "Delhi",
    "Pune",
    "Kochi"
]

destinations = [
    "Chennai",
    "Coimbatore",
    "Bangalore",
    "Hyderabad",
    "Mumbai",
    "Delhi",
    "Pune",
    "Kochi"
]

shipping_modes = [
    "Road",
    "Rail",
    "Air",
    "Sea"
]

weather_conditions = [
    "Clear",
    "Rain",
    "Storm",
    "Fog"
]

traffic_conditions = [
    "Low",
    "Medium",
    "High"
]

data = []

for i in range(1, n + 1):

    vendor = random.choice(vendors)
    origin = random.choice(origins)
    destination = random.choice(destinations)

    while destination == origin:
        destination = random.choice(destinations)

    shipping_mode = random.choice(shipping_modes)

    distance = random.randint(50, 2500)

    weather = random.choices(
        weather_conditions,
        weights=[50, 30, 15, 5]
    )[0]

    traffic = random.choices(
        traffic_conditions,
        weights=[45, 35, 20]
    )[0]

    planned_days = max(
        1,
        int(distance / random.randint(250, 500))
    )

    delay = 0

    # Weather effect
    if weather == "Rain":
        delay += random.choice([0, 1, 1, 2])

    elif weather == "Storm":
        delay += random.choice([1, 2, 3, 4])

    elif weather == "Fog":
        delay += random.choice([0, 1, 2])

    # Traffic effect
    if traffic == "Medium":
        delay += random.choice([0, 1])

    elif traffic == "High":
        delay += random.choice([1, 2, 3])

    # Shipping mode effect
    if shipping_mode == "Road":
        delay += random.choice([0, 0, 1])

    elif shipping_mode == "Sea":
        delay += random.choice([0, 1, 2])

    # Random variation
    delay += random.choice([0, 0, 0, 1])

    actual_days = planned_days + delay

    delayed = 1 if actual_days > planned_days else 0

    data.append([
        f"SHP{i:04d}",
        vendor,
        origin,
        destination,
        shipping_mode,
        distance,
        weather,
        traffic,
        planned_days,
        actual_days,
        delay,
        delayed
    ])

shipping_df = pd.DataFrame(
    data,
    columns=[
        "Shipment_ID",
        "Vendor",
        "Origin",
        "Destination",
        "Shipping_Mode",
        "Distance_km",
        "Weather",
        "Traffic",
        "Planned_Delivery_Days",
        "Actual_Delivery_Days",
        "Delay_Days",
        "Delayed"
    ]
)

shipping_df.head()
```

```python
contract_data = []

for vendor in vendors:

    delivery_period = random.randint(5, 15)

    payment_period = random.choice([
        15, 30, 45, 60
    ])

    delay_notification = random.choice([
        12, 24, 24, 48, 72
    ])

    insurance = random.choice([
        "Present",
        "Present",
        "Present",
        "Missing"
    ])

    confidentiality = random.choice([
        "Present",
        "Present",
        "Present",
        "Missing"
    ])

    termination = random.choice([
        "Present",
        "Present",
        "Present",
        "Missing"
    ])

    contract_data.append([
        vendor,
        delivery_period,
        payment_period,
        delay_notification,
        insurance,
        confidentiality,
        termination
    ])

contract_df = pd.DataFrame(
    contract_data,
    columns=[
        "Vendor",
        "Delivery_Period_Days",
        "Payment_Period_Days",
        "Delay_Notification_Hours",
        "Insurance",
        "Confidentiality",
        "Termination_Clause"
    ]
)

contract_df
```

```python
contract_df["Compliance_Score"] = (
    (
        (contract_df["Insurance"] == "Present").astype(int) +
        (contract_df["Confidentiality"] == "Present").astype(int) +
        (contract_df["Termination_Clause"] == "Present").astype(int)
    ) / 3 * 100
).round(2)

contract_df
```

```python
def get_compliance_status(score):

    if score == 100:
        return "Compliant"

    elif score >= 66.67:
        return "Partially Compliant"

    else:
        return "Non-Compliant"


contract_df["Compliance_Status"] = (
    contract_df["Compliance_Score"]
    .apply(get_compliance_status)
)

contract_df
```

```python
contract_df.to_csv(
    "vendor_contracts.csv",
    index=False
)

print("Random vendor contract database created successfully!")
```

```python
import pandas as pd
import numpy as np

shipping_df = pd.read_csv("shipping_data.csv")
contract_df = pd.read_csv("vendor_contracts.csv")

print("Shipping Database:", shipping_df.shape)
print("Contract Database:", contract_df.shape)

display(shipping_df.head())
display(contract_df.head())
```

```python
print("Shipping Dataset Information")
print("=" * 40)

print("Rows:", len(shipping_df))
print("Columns:", len(shipping_df.columns))

print("\nMissing Values:")
print(shipping_df.isnull().sum())

print("\nDelayed Distribution:")
print(shipping_df["Delayed"].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))

sns.countplot(
    x="Delayed",
    data=shipping_df
)

plt.title("Supply Chain Delay Distribution")
plt.xlabel("Delayed (0 = No, 1 = Yes)")
plt.ylabel("Number of Shipments")

plt.show()
```

```python
features = [
    "Vendor",
    "Origin",
    "Destination",
    "Shipping_Mode",
    "Distance_km",
    "Weather",
    "Traffic",
    "Planned_Delivery_Days"
]

target = "Delayed"

X = shipping_df[features]
y = shipping_df[target]

print("Input Features:")
print(features)

print("\nTarget:")
print(target)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
```

```python
categorical_features = [
    "Vendor",
    "Origin",
    "Destination",
    "Shipping_Mode",
    "Weather",
    "Traffic"
]

numerical_features = [
    "Distance_km",
    "Planned_Delivery_Days"
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessing pipeline created.")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", model)
])

print("Random Forest model created.")
```

```python
pipeline.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
y_pred = pipeline.predict(X_test)

print("Predictions generated!")

print("\nFirst 20 predictions:")
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        target_names=[
            "On Time",
            "Delayed"
        ]
    )
)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["On Time", "Delayed"],
    yticklabels=["On Time", "Delayed"]
)

plt.title("Supply Chain Delay Prediction")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
prediction_results = X_test.copy()

prediction_results["Actual_Delayed"] = y_test.values
prediction_results["Predicted_Delayed"] = y_pred

prediction_results["Actual_Status"] = (
    prediction_results["Actual_Delayed"]
    .map({
        0: "On Time",
        1: "Delayed"
    })
)

prediction_results["Predicted_Status"] = (
    prediction_results["Predicted_Delayed"]
    .map({
        0: "On Time",
        1: "Delayed"
    })
)

prediction_results.head(10)
```

```python
probability = pipeline.predict_proba(X_test)

prediction_results["Delay_Probability"] = (
    probability[:, 1] * 100
).round(2)

prediction_results.head(10)
```

```python
high_risk_shipments = prediction_results[
    prediction_results["Predicted_Delayed"] == 1
].sort_values(
    "Delay_Probability",
    ascending=False
)

print(
    "Number of predicted delayed shipments:",
    len(high_risk_shipments)
)

display(high_risk_shipments.head(20))
```

```python
vendor_prediction = (
    prediction_results
    .groupby("Vendor")
    .agg(
        Predicted_Delayed=("Predicted_Delayed", "max"),
        Average_Delay_Probability=("Delay_Probability", "mean"),
        Shipment_Count=("Vendor", "count")
    )
    .reset_index()
)

vendor_prediction["Average_Delay_Probability"] = (
    vendor_prediction["Average_Delay_Probability"]
    .round(2)
)

vendor_prediction
```

```python
combined_data = pd.merge(
    vendor_prediction,
    contract_df,
    on="Vendor",
    how="left"
)

combined_data
```

```python
def calculate_risk(row):

    delayed = row["Predicted_Delayed"]
    compliance = row["Compliance_Status"]

    if delayed == 1 and compliance == "Non-Compliant":
        return "High Risk"

    elif delayed == 1:
        return "Medium Risk"

    elif compliance == "Non-Compliant":
        return "Medium Risk"

    else:
        return "Low Risk"


combined_data["Overall_Risk"] = (
    combined_data.apply(
        calculate_risk,
        axis=1
    )
)

combined_data[
    [
        "Vendor",
        "Predicted_Delayed",
        "Compliance_Status",
        "Overall_Risk"
    ]
]


```

```python
final_output = combined_data[
    [
        "Vendor",
        "Shipment_Count",
        "Predicted_Delayed",
        "Average_Delay_Probability",
        "Compliance_Score",
        "Compliance_Status",
        "Overall_Risk"
    ]
].copy()

final_output["Predicted_Status"] = (
    final_output["Predicted_Delayed"]
    .map({
        0: "On Time",
        1: "Delayed"
    })
)

final_output = final_output[
    [
        "Vendor",
        "Shipment_Count",
        "Predicted_Status",
        "Average_Delay_Probability",
        "Compliance_Score",
        "Compliance_Status",
        "Overall_Risk"
    ]
]

display(final_output)
```
